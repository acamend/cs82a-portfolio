Given problem: Write a function that multiplies two whole numbers WITHOUT using the * operator: multiply by adding one number to itself repeatedly inside a loop. Do it once with a for loop and once with a while loop, and show both agree with the real product.

Thought Process: If we want to multiply 4x5, set X = 4 and Y=5. Write a loop that adds x to itself and runs the loop y times. 4+4+4+4+4. Both answers will result in 20.

So we will set x+x inside a loop that runs y times. So there must be a counter that runs the loop until the count equals (==) y. Start with count=0, inside the loop count = count+1. Run until count ==y.

Do once with a "while" loop and once with a "for" loop. We can set x and y to any two whole number we choose. For this assignment it is 6x7 = 42.

"While" loop

```python
# This code calculates x*y using a while loop. Please enter the x and y you desire.

x=6
y=7
sum=0
count=0
while count<=y:
    sum = sum + x
    count += 1
    if count == y:
        print(f"{x} * {y} = {sum}")
        result = (int(sum) == int(x*y))
        print(result)
```

"For" loop

```python
# This code calculates x*y using a for loop. Please enter the x and y you desire.

x=6
y=7
count = 0
sum=0

for count in range (0,int(y)):
    sum = sum + x
    count += 1
    if count == int(y):
        print(f"{x} * {y} = {sum}")
        result = (int(sum) == int(x*y))
        print(result)
```

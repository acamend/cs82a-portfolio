Given Problem: 
Given a list of temperatures, write a python script that categorizes each as cold, warm, or hot.
Cold = below 15C.
Warm = 15-25C (inclusive).
Hot = above 25C.
Temps = [10, 16, 23, 28, 14, 22, 20]

Additional information: use if/elif/else inside a loop, and print each temperature with its label. 

Thought Process:
I need to define the temperature dictionary and then run the dictionary through a loop that will read each temperature, run it through the series of if statements to determine temperature category and then print each temperature and category.

```python
# This code categories a list of temperatures entered in Celcius as cold, warm or hot.

temps = [10, 16, 23, 28, 14, 22, 20, 25, 15]
for x in temps:
    if x > 25:
        print (x,"Hot")
    elif x < 15:
        print (x, "Cold")
    else:
        print (x, "Warm")
```

# T-Test on feed and chick weight

```python
#data set mean
df.groupby("feed")["weight"].mean().sort_values()
```

```python
#data set median
df.groupby("feed")["weight"].median().sort_values()
```

```python
#count of values in each category
df.groupby("feed")["weight"].count().sort_values()
```

Null assumption: Feed type has no effect on chick weight. Compare horsebean and casein weights.

```python
from scipy.stats import ttest_ind
a = df[df["feed"] == "casein"]["weight"]
b = df[df["feed"] == "horsebean"]["weight"]
ttest_ind(a, b)
```

```python
df.boxplot(column = 'weight', by= 'feed')
```

The mean weight of horsebean fed chicks is 160.2 grams. The mean weight of casein fed chicks is 328.92 grams. The very small p value (8.25e-07) suggests that it would be unlikely to observe this difference in weight if the null hypothesis were true. One caution, there are only about 10 data points in each sample set. A larger sample size would provde a more confident result.

# Step 1

```python
# Teach and Test using a 70/30 split
from sklearn.model_selection import train_test_split
X = df[["total_bill"]]
y = df["tip"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
#asked AI how to view to see if anything happened
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

# Step 2
We are making a model to predict the tip amount based in the total bill. We have selected to train the model on 70 percent of the data and test it on 30 percent. We must split the data into two groups so the model does not lean all the answers by using all the data. We need to give the model enough data to be able to perform a regression but we must tes the model on data it has never seen before. If the model used all the data to train, we could not verify that it works without feeding it unseen data.

# Step 3 - Fit the regression and read the slope

```python
# Created and Fit
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

```python
#Just looking at the train score for fun
print(model.score(X_train, y_train))
```

```python
# Slope (model.coefficient) and intercept (model.intercept)
print(model.coef_, model.intercept_)
```

# Model Prediction
Based on the regression, the prediction is that for each dollar of the total bill, there will be an 11 cent tip.

```python
# Split the data while ensuring all three target classes are proportionally represented
iris_training, iris_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.3, 
    random_state=357, 
    stratify=y  # <--- This ensures all three outcomes are split evenly!
)

# Convert labels to factorized numeric arrays
y_train = pd.factorize(y_train)[0]
y_test = pd.factorize(y_test)[0]
```

```python
#Confirm split
print("X_train shape:", iris_training.shape)
print("y_train shape:", y_train.shape)
```

# Step 6: Fit and score kNN

```python
#Assignment text copied here
from sklearn.neighbors import KNeighborsClassifier
from sklearn import tree #added based on video viewed
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(iris_training, y_train)
```

```python

from sklearn.metrics import accuracy_score, confusion_matrix
y_pred = knn.predict(iris_test)

unique, frequency = np.unique(y_pred, return_counts = True)
print(unique)
print(frequency)
```

```python
# Accuracy test (from video)
from sklearn.metrics import accuracy_score, confusion_matrix
kNN_accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy of the model:{kNN_accuracy*100:.2f}%")
```

```python
#Check acccuracy with pandas (from book)
df = pd.DataFrame({'iris_pred': y_pred, 'iris.testLabels':y_test})
cross_tab = pd.crosstab(df['iris_pred'], df['iris.testLabels'], margins=True)
print(cross_tab)
```

The accuraacy of the kNN model is 98%. Of the 45 predictions, the model identified one target incorrectly. It named a viginica species as versicolor.

# Step 7: Fit, score, and read a shallow tree

```python
#Import tree and fit data
from sklearn.tree import DecisionTreeClassifier
tree = DecisionTreeClassifier(max_depth=3)

tree.fit(iris_training,y_train) #AI helped add this line

#Program predictor
y_pred_tree = tree.predict(iris_test)

#Used AI for the following code, not in book for tree (only for kNN)

# Calculate accuracy score
tree_accuracy = accuracy_score(y_test, y_pred_tree)

# Print out the results
print(f"Decision Tree Accuracy: {tree_accuracy * 100:.2f}%")
```

```python
#Print shallow tree scores
print(f"Decision Tree Accuracy of Train: {tree.score(iris_training, y_train)}")

print(f"Decision Tree Accuracy of Test: {tree.score(iris_test, y_test)}")
```

```python
from sklearn.tree import plot_tree

#This line was amended with assistance from the video to get the target names added
plot_tree(tree, feature_names=iris_sklearn.feature_names, class_names=iris_sklearn.target_names, filled=True)
plt.show()
```

The model has determined that pedal length is the first classification step. If the petal length is less than or equal to 2.6 cm, the iris is the setosa species. If the petal length is greater thatn 2.6 cm, the petal width is checked. If the petal width is greater than 1.75 cm, the iris is classified as virginica. If the petal width is less than or equal to 1.75 cm, then we must check the petal length again. If the petal length is great than 4.95, the species is virginica, if not, it is versicolor.

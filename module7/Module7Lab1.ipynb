# Step 1

```python
# Teach and Test using a 70/30 split
from sklearn.model_selection import train_test_split
X = df[["total_bill"]]
y = df["tip"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
#asked AI how to view to see if anything happened
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

# Step 2
We have selected to train the model on 70 percent of the data and test it on 30 percent. We must split the data into two groups so the model does not lean all the answers by using all the data. We need to give the model enough data to be able to perform a regression but we must tes the model on data it has never seen before. If the model used all the data to train, we could not verify that it works without feeding it unseen data.

```python
#Step 3
```

```python
# Created, Fit, Score
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

```python
#Just looking at the train score for fun
print(model.score(X_train, y_train))
```

```python
# Slope (model.coefficient) and intercept (model.intercept)
print(model.coef_, model.intercept_)
```

# Model Prediction
Based on the regression, the prediction is that for each dollar of the total bill, there will be an 11 cent tip.

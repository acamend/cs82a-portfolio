```python
df["difference"] = df["Y"] - df["X"]
df["percentchange"] = df["difference"]/df["X"]*100
```

```python
print(f"Mean: {df["percentchange"].mean()}")
print(f"Median: {df["percentchange"].median()}")
print(f"Min: {df["percentchange"].min()}")
print(f"Max: {df["percentchange"].max()}")
```

```python
#Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
```

```python
#REBUILD Lab 2 Table Here
#Connect to SQL3 database
con = sqlite3.connect("store.db")




con.commit() #Do this after every addition!
```

```python
#Create cursor to execute SQL communications
cur = con.cursor()
```

```python
# Drop table before creating so no duplicates
cur.execute("""
DROP TABLE IF EXISTS 'customers';""")
```

```python
#Create a table - with at least id, name, city, and one numeric column and 8 rows
cur.execute("""
CREATE TABLE IF NOT EXISTS 'customers' (
'ID' INTEGER PRIMARY KEY AUTOINCREMENT,
'Name' TEXT,
'City' TEXT,
'NumInHousehold' INTEGER);""")
```

```python
#Insert values into customer table
cur.execute("""
    INSERT INTO customers (Name, City, NumInHousehold) VALUES
        ("Andrew", "Anchorage", 1),
        ("Betty", "Boston", 2),
        ("Charles", "Chatanooga", 3),
        ("Darla", "Detroit", 4),
        ("Elliot", "El Paso", 1),
        ("Frances", "Frisco", 2),
        ("Gerald", "Galveston", 3),
        ("Henrietta", "Houston", 4);""")
```

```python
#Check input
pd.read_sql("SELECT * FROM customers", con)
```

```python
#Query all data for customers having households greater than 1 person (SELECT WHERE)
cur.execute("""
    SELECT *
    FROM customers
    WHERE NumInHousehold > 1;
    """)
#Fetch all matching rows
rows = cur.fetchall()

# Print the results
print("Customers with more than 1 person in the household:")
for row in rows:
    print(row)
```

```python
#Query all data for customers having households greater than 1 person and order by name
cur.execute("""
    SELECT *
    FROM customers
    WHERE NumInHousehold > 1
    ORDER BY Name;
    """)

#Fetch all matching rows
rows = cur.fetchall()

# Print the results
print("Customers with more than 1 person in the household, ordered by name:")
for row in rows:
    print(row)
```

```python
#Query the customer database for customers living in cities containing "San", (LIKE San Diego and San Francisco)

cur.execute("""
    SELECT *
    FROM customers
    WHERE City LIKE '%on%';
    """)

#Fetch all matching rows
rows = cur.fetchall()

# Print the results
print("Customers living in cities containing '%on':")
for row in rows:
    print(row)
```

```python
#Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
```

```python
#Connect to SQL3 database
con = sqlite3.connect("store.db")
```

```python
#Create cursor to execute SQL communications
cur = con.cursor()
```

```python
# Drop table before creating so no duplicates
cur.execute("""
DROP TABLE IF EXISTS 'customers';""")
```

```python
#Create a table - with at least id, name, city, and one numeric column and 8 rows
cur.execute("""
CREATE TABLE IF NOT EXISTS 'customers' (
'ID' INTEGER PRIMARY KEY AUTOINCREMENT,
'Name' TEXT,
'City' TEXT,
'NumInHousehold' INTEGER);""")
# Creating autoincrement code for 'ID' restated by Anaconda Assistant, different from book.
```

```python
#Insert values into customer table
cur.execute("""
    INSERT INTO customers (Name, City, NumInHousehold) VALUES
        ("Andrew", "Anchorage", 1),
        ("Betty", "Boston", 2),
        ("Charles", "Chatanooga", 3),
        ("Darla", "Detroit", 4),
        ("Elliot", "El Paso", 1),
        ("Frances", "Frisco", 2),
        ("Gerald", "Galveston", 3),
        ("Henrietta", "Houston", 4);""")
```

```python
#Check first 3 rows
pd.read_sql("SELECT * FROM customers", con)
```

```python
for row in cur.execute("SELECT Name, City FROM customers ORDER BY City"):
    print(row)
```
